```python
# Once you've run the pipeline a few times, query the historical data back out
from db import get_connection
import pandas as pd

conn = get_connection()
history_df = pd.read_sql(
    """
    SELECT c.name, r.recorded_at, r.temperature_c, r.humidity_pct
    FROM readings r
    JOIN cities c ON c.city_id = r.city_id
    ORDER BY r.recorded_at DESC
    LIMIT 50
    """,
    conn,
)
conn.close()
history_df
```

```python
# Quick trend plot once you have enough historical rows
history_df.plot(x='recorded_at', y='temperature_c', kind='line', figsize=(10, 4))
```
